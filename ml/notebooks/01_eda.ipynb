# 01. Exploratory Data Analysis (EDA) - Telco Customer Churn

This notebook explores the IBM Telco Customer Churn dataset to uncover key patterns, distributions, and correlations driving customer attrition.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="muted")
%matplotlib inline

data_path = os.path.join("..", "data", "customer_churn.csv")
df = pd.read_csv(data_path)
print(f"Dataset Shape: {df.shape}")
df.head()
```

## 1. Target Variable Distribution (Churn)

```python
churn_counts = df['Churn'].value_counts()
print("Churn Distribution:")
print(df['Churn'].value_counts(normalize=True))

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Churn', hue='Churn', palette=['#3b82f6', '#ef4444'], legend=False)
plt.title('Customer Churn Class Distribution (Imbalanced ~26.5% Churn)')
plt.show()
```

## 2. Tenure vs. Churn Analysis

```python
plt.figure(figsize=(10, 4))
sns.histplot(data=df, x='tenure', hue='Churn', kde=True, bins=30, palette=['#3b82f6', '#ef4444'])
plt.title('Tenure Distribution by Churn Status (High risk in months 0-12)')
plt.xlabel('Tenure (Months)')
plt.show()
```

## 3. Contract Type vs. Churn Analysis

```python
plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='Contract', hue='Churn', palette=['#3b82f6', '#ef4444'])
plt.title('Contract Type Impact on Churn (Month-to-month dominates churn)')
plt.show()
```

## 4. Monthly Charges and Internet Service

```python
plt.figure(figsize=(10, 4))
sns.boxplot(data=df, x='InternetService', y='MonthlyCharges', hue='Churn', palette=['#3b82f6', '#ef4444'])
plt.title('Monthly Charges by Internet Service and Churn (Fiber optic has highest churn)')
plt.show()
```
